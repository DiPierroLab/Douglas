```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
```

# Unpaired

```python
# The energy loops are 4 beads by 4 beads
L413 = np.loadtxt('/Users/douglas/Documents/Features_Transfer/store_lambdas/lambdas413.txt',delimiter = ',')
plt.matshow(L413[249:255,749:755])
plt.title('The energy loops are 4 beads by 4 beads')
```

```python
M413 = np.loadtxt('/Users/douglas/Documents/Features_Transfer/store_Hi-C/Hi-C_directory_413.txt')
```

```python
# Zoomed out view
plt.matshow(M413, norm=mpl.colors.LogNorm(vmin=0.002, vmax=0.03),cmap="Reds")
plt.title('413 Unpaired')
plt.colorbar()
```

```python
# Zoomed in view
plt.matshow(M413[0:1000,0:1000], norm=mpl.colors.LogNorm(vmin=0.002, vmax=0.03),cmap="Reds")
plt.title('413 Unpaired; zoomed in paternal')
plt.colorbar()
```

```python
# Zoomed in further
plt.matshow(M413[200:300,700:800], norm=mpl.colors.LogNorm(vmin=0.002, vmax=0.03),cmap="Reds")
plt.title('413 Unpaired; zoomed further onto crosshair peak')
plt.colorbar()
```

```python
# Zoomed in view of diagonal
plt.matshow(M413[700:800,700:800], norm=mpl.colors.LogNorm(vmin=0.002, vmax=0.03),cmap="Reds")
plt.title('413 Unpaired; zoomed further onto paternal diagonal below crosshair peak')
plt.colorbar()
```
